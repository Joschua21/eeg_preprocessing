# EDF Trimming Utility
Provide an EDF path and a time window in seconds to create a trimmed file.

```python
from pathlib import Path
import mne

edf_path = Path(r"C:\Users\HarrisLab\Desktop\temp\sub-053_ses-011_recording-001.edf")
start_s = 0.0
end_s = 84000.0

if end_s <= start_s:
    raise ValueError("end_s must be greater than start_s")

raw = mne.io.read_raw_edf(edf_path, preload=True, verbose="ERROR")
trimmed = raw.copy().crop(tmin=start_s, tmax=end_s)

out_path = edf_path.with_name(f"{edf_path.stem}_trimmed{edf_path.suffix}")
trimmed.export(out_path, fmt="edf", overwrite=True)
print(f"Saved trimmed EDF: {out_path}")
```
