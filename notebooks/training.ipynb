# Somnotate Training
Run the full training workflow using .mat files and save the trained model.

```python
%load_ext autoreload
%autoreload 2

from pathlib import Path
from utils.training import run_training
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
model_name = "test_model_1"  # required
repo_root = Path.cwd().parent
training_mat_dir = repo_root / "data" / "hypnose_eeg" / "somnotate_training"

model_path = run_training(
    model_name=model_name,
    training_mat_dir=training_mat_dir,
    repo_root=repo_root,
)

print("Saved model to:", model_path)
```

```python
results_dir = (
    repo_root
    / "data"
    / "hypnose_eeg"
    / "derivatives"
    / "somnotate_training"
    / model_name
    / "results"
 )
evaluation_csv = results_dir / "evaluation_by_recording.csv"
evaluation_npz = results_dir / "evaluation.npz"

if evaluation_csv.exists():
    df = pd.read_csv(evaluation_csv)
    print("Hold-out accuracies:")
    for _, row in df.iterrows():
        print(f"{row['recording']}: {row['accuracy'] * 100:.2f}%")
elif evaluation_npz.exists():
    data = np.load(evaluation_npz)
    accuracies = data.get("accuracy")
    if accuracies is None:
        print("No accuracy array in evaluation.npz.")
    else:
        print("Hold-out accuracies (order matches training input):")
        for idx, value in enumerate(accuracies, start=1):
            print(f"recording_{idx}: {value * 100:.2f}%")
else:
    print("No evaluation_by_recording.csv or evaluation.npz found.")

if evaluation_npz.exists():
    data = np.load(evaluation_npz)
    confusion = data["confusion"]
    labels = ["U", "W", "N", "R"]

    fig, ax = plt.subplots(figsize=(6, 5))
    im = ax.imshow(confusion, cmap="Blues")
    ax.set_xticks(range(4))
    ax.set_yticks(range(4))
    ax.set_xticklabels(labels)
    ax.set_yticklabels(labels)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("Manual")

    totals = confusion.sum(axis=1, keepdims=True)
    confusion_pct = np.divide(
        confusion,
        totals,
        out=np.zeros_like(confusion, dtype=float),
        where=totals != 0,
    )
    for i in range(confusion_pct.shape[0]):
        for j in range(confusion_pct.shape[1]):
            ax.text(j, i, f"{confusion_pct[i, j]:.2f}", ha="center", va="center", color="black")

    fig.colorbar(im, ax=ax, shrink=0.8)
    fig.tight_layout()
    plt.show()
else:
    print("No evaluation.npz found.")
```
